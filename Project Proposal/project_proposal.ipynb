# Research Objective

*Requires data modeling and quantitative research in Transport, Infrastructure & Logistics*

We would like to research the speed limit change from 50km/h to 30km/h in Amsterdam. On many roads in the city the speed limit was changed. We are curious to see how this change has affected the amount of traffic accidents in Amsterdam. Therefore we propose the following research question, along with some subquestions that we will first answer. Probably not all, but this is just the first list of questions that we found interesting. Later we will make a final decision on what questions we will research.

How has the change from the speed limit of 50kmh to 30kmh affected the amount and severity of traffic accidents in Amsterdam?
1) Has the severity of the traffic accidents in Amsterdam changed since the change in speed limit?
 2) Has the amount of traffic fatalities in Amsterdam changed after the change in speed limit?
3) How does this compare to other large Dutch cities that have not changed the speed limit?
4) Is there a difference in effect between types of vehicles? (bicycles, cars, motorcycles)
5) What is the difference between the predicted trend of traffic accidents in Amsterdam from 2023-now (if there was no intervention) compared to the real data?
6) Is there a greater effect on the roads with the changed speed limit from 50-30km/h and the roads without change?
7) Is there a difference in effect on different times of the day and on different days of the week?
8) What is the effect of weather circumstances?
9) What is the effect between different types of roads/locations? e.g. traffic accidents in the inner and outer city?

# Data Pipeline

We use two main types of data: traffic accident reports and the speed limit per road segment. Using these we can determine the volume of traffic accidents and compare and contrast between affected roads before and after intervention, and between roads that have and have not been changed.

BRON datasets include incidents across the entire country, so we filter by Amsterdam and the control cities. They also include many details such as the types of vehicles involved, type of road segment, type of road surface, time of day, weather conditions, etc. These can all be compared to investigate where the intervention has been most effective. (And we can filter out the irrelevant bits.)

Using the historical data, we can make a prediction how traffic accidents might have changed if the speed limit was not lowered and compare this to the actual situation. This can then also be compared to the actual and predicted trends in the control cities.

We then visualize our results with various block charts, pie charts, and graphs to compare between different categories of road segments and accident details. And we can generate (animated) heatmaps to illustrate how different parts of the selected cities are affected.
