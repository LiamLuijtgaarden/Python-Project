```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df_amsterdam = pd.read_csv("ongevallen_gefilterd_amsterdam.csv")
df_landelijk = pd.read_csv("ongevallen_gefilterd_landelijk.csv")

bevolking = pd.read_csv("bevolking_nederland_steden.csv", 
                        sep=";",
                        skiprows=7)

bevolking["Perioden"] = pd.to_numeric(
    bevolking["Perioden"], errors="coerce"
).astype("Int64")

aantal = (
    bevolking["aantal"]
    .astype(str)
    .str.strip()
    .str.replace(".", "", regex=False)   
    .str.replace(",", ".", regex=False) 
)

bevolking["aantal"] = pd.to_numeric(aantal, errors="coerce")

df_utrecht = df_landelijk[df_landelijk["GME_NAAM"] == "Utrecht"].copy()
df_denhaag = df_landelijk[df_landelijk["GME_NAAM"] == "'s-Gravenhage"].copy()
df_rotterdam = df_landelijk[df_landelijk["GME_NAAM"] == "Rotterdam"].copy()
df_eindhoven = df_landelijk[df_landelijk["GME_NAAM"] == "Eindhoven"].copy()
df_groningen = df_landelijk[df_landelijk["GME_NAAM"] == "Groningen"].copy()

display(df_amsterdam.head())
display(df_landelijk.head())

print(bevolking.columns)
print(bevolking.dtypes)
display(bevolking.head())
```

DF Landelijk

```python
print("Accidents per year:")
display(df_landelijk["JAAR_VKL"].value_counts().sort_index())

# Verdeling van de ernstcategorieën
print("Severity categories:")
display(df_landelijk["AP3_CODE"].value_counts(dropna=False))

# Controle op dubbele ongevallen
print("Duplicate accident IDs:")
print(df_landelijk["VKL_NUMMER"].duplicated().sum())

# Tel ongevallen per jaar en ernstcategorie
severity_yearly_landelijk = pd.crosstab(
    df_landelijk["JAAR_VKL"],
    df_landelijk["AP3_CODE"]
)

# Zorg voor vaste kolomvolgorde
categories = ["DOD", "LET", "UMS"]

severity_yearly_landelijk = severity_yearly_landelijk.reindex(
    columns=categories,
    fill_value=0
)

# Totaal per jaar
severity_yearly_landelijk["Total"] = severity_yearly_landelijk.sum(axis=1)

display(severity_yearly_landelijk)
```

Final Tables

```python
def add_per_capita_rates(severity_table, region_name):
    result = severity_table.copy()

    population = bevolking.loc[
        bevolking["Regio's"] == region_name, ["Perioden", "aantal"]
    ].copy()

    population["Perioden"] = pd.to_numeric(population["Perioden"], errors="coerce")
    population["aantal"] = pd.to_numeric(population["aantal"], errors="coerce")
    population = population.dropna().set_index("Perioden")["aantal"]

    result.index = pd.to_numeric(result.index)
    population_for_year = population.reindex(result.index)

    for category in ["DOD", "LET", "UMS", "Total"]:
        result[f"{category}_per_100k"] = (
            result[category] / population_for_year * 100_000
        )

    return result

severity_yearly_amsterdam_capita = add_per_capita_rates(
    severity_yearly_amsterdam, "Amsterdam"
)
severity_yearly_utrecht_capita = add_per_capita_rates(
    severity_yearly_utrecht, "Utrecht (gemeente)"
)
severity_yearly_denhaag_capita = add_per_capita_rates(
    severity_yearly_denhaag, "'s-Gravenhage (gemeente)"
)
severity_yearly_rotterdam_capita = add_per_capita_rates(
    severity_yearly_rotterdam, "Rotterdam"
)
severity_yearly_eindhoven_capita = add_per_capita_rates(
    severity_yearly_eindhoven, "Eindhoven"
)
severity_yearly_groningen_capita = add_per_capita_rates(
    severity_yearly_groningen, "Groningen (gemeente)"
)
severity_yearly_landelijk_capita = add_per_capita_rates(
    severity_yearly_landelijk, "Nederland"
)

print("Landelijk:")
display(severity_yearly_landelijk_capita)
print("Amsterdam:")
display(severity_yearly_amsterdam_capita)
print("Rotterdam:")
display(severity_yearly_rotterdam_capita)
print("Den Haag:")
display(severity_yearly_denhaag_capita)
print("Utrecht:")
display(severity_yearly_utrecht_capita)
print("Eindhoven:")
display(severity_yearly_eindhoven_capita)
print("Groningen:")
display(severity_yearly_groningen_capita)
```
